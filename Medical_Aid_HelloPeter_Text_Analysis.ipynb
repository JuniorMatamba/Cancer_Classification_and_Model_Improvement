```python
from collections import Counter
import re
import pandas as pd

# Combine all cleaned reviews into one text
all_words = " ".join(df_clean['cleaned_text'].dropna())

# Split the text into individual words
words = re.findall(r'\b[a-zA-Z]+\b', all_words.lower())

# Remove common words that do not add much meaning
stop_words = {
    'the', 'and', 'for', 'are', 'was', 'were', 'with',
    'this', 'that', 'have', 'has', 'had', 'you', 'your',
    'but', 'not', 'from', 'they', 'them', 'their', 'there',
    'been', 'being', 'would', 'could', 'should', 'will',
    'can', 'all', 'our', 'out', 'get', 'got', 'just',
    'very', 'one', 'about', 'what', 'when', 'where'
}

# Keep meaningful words
filtered_words = [
    word for word in words
    if word not in stop_words and len(word) > 2
]

# Count the frequency of each word
word_counts = Counter(filtered_words)

# Select the 20 most frequent words
top_words = word_counts.most_common(20)

# Convert the results into a DataFrame
top_words_df = pd.DataFrame(
    top_words,
    columns=['Word', 'Frequency']
)

# Display the results
display(top_words_df)
```

Now Converting text into numerical features using TF-IDF
We now move from data exploration to feature extraction.

Machine learning algorithms require numerical inputs. TF-IDF converts the review text into numerical values based on the importance of words across the collection of reviews.

```python
from sklearn.feature_extraction.text import TfidfVectorizer

print("TfidfVectorizer imported successfully!")
```

```python
# STEP 8: APPLY TF-IDF VECTORIZATION

tfidf_vectorizer = TfidfVectorizer(
    max_features=1000,
    ngram_range=(1, 2),
    min_df=1,
    max_df=0.95
)

# Convert cleaned reviews into numerical features
tfidf_matrix = tfidf_vectorizer.fit_transform(
    df_clean['cleaned_text']
)

# Display the dimensions of the resulting matrix
print("TF-IDF matrix shape:", tfidf_matrix.shape)

# Display some of the extracted words
print("\nFirst 20 extracted features:")
print(tfidf_vectorizer.get_feature_names_out()[:20])
```

```python

# STEP 8.1: DISPLAY TF-IDF VOCABULARY

feature_names = tfidf_vectorizer.get_feature_names_out()

print("First 30 extracted features:")

print(feature_names[:30])
```

```python

# STEP 8.2: DISPLAY A SMALL PART OF THE TF-IDF MATRIX

tfidf_preview = pd.DataFrame(
    tfidf_matrix[:5].toarray(),
    columns=feature_names
)

display(tfidf_preview.iloc[:, :15])
```

This displays the numerical representation of selected terms in the first five reviews.

Topic modelling using NMF
Now we use Non-negative Matrix Factorisation to identify groups of terms that may represent recurring customer concerns.

We will start with three topics. This is an exploratory choice, not a claim that the dataset necessarily contains exactly three genuine complaint categories

```python
from sklearn.decomposition import NMF

print("NMF imported successfully!")
```

```python
# STEP 9: APPLY NON-NEGATIVE MATRIX FACTORIZATION (NMF)

# Set the number of topics to discover
number_of_topics = 3

# Create the NMF model
nmf_model = NMF(
    n_components=number_of_topics,
    random_state=42,
    init='nndsvda',
    max_iter=500
)

# Train the NMF model using the TF-IDF matrix
W = nmf_model.fit_transform(tfidf_matrix)

# Extract the topic-word components
H = nmf_model.components_

print("NMF model trained successfully!")
print("Document-topic matrix shape:", W.shape)
print("Topic-word matrix shape:", H.shape)
```

```python

# STEP 9.2: DISPLAY THE TOP WORDS IN EACH TOPIC

feature_names = tfidf_vectorizer.get_feature_names_out()

for topic_index, topic in enumerate(H):

    print("\nTOPIC", topic_index + 1)

    # Get the indexes of the 10 highest-weighted words
    top_word_indexes = topic.argsort()[-10:][::-1]

    # Display the corresponding words
    top_terms = [
        feature_names[index]
        for index in top_word_indexes
    ]

    print(", ".join(top_terms))
```

You may see groups of words associated with different experiences. Examine each group and then go back to the original reviews to determine what the topic actually represents.

Do not automatically label Topic 1 as claims or Topic 2 as customer service. The topic names must be supported by the actual output.

```python

# STEP 9.3: ASSIGN THE MOST LIKELY TOPIC TO EACH REVIEW

df_clean['topic'] = W.argmax(axis=1)

# Display topic assignments
display(
    df_clean[['cleaned_text', 'topic']].head(10)
)

# Count the reviews assigned to each topic
print("\nNumber of reviews in each topic:")

print(
    df_clean['topic']
    .value_counts()
    .sort_index()
)
```

```python

# STEP 9.4: VISUALISE TOPIC DISTRIBUTION

topic_counts = (
    df_clean['topic']
    .value_counts()
    .sort_index()
)

plt.figure(figsize=(8, 5))

sns.barplot(
    x=topic_counts.index + 1,
    y=topic_counts.values
)

plt.title('Distribution of Reviews Across NMF Topics')
plt.xlabel('Topic')
plt.ylabel('Number of Reviews')

plt.show()
```

Prepare sentiment categories
We will now create sentiment labels from the original customer ratings.

The original distribution is:
- Negative: ratings 1 and 2.
- Neutral: rating 3.
- Positive: ratings 4 and 5.
Because the neutral and positive classes contain very few records, we will combine them for a cautious binary classification experiment.

```python

# STEP 10.1: CREATE SENTIMENT LABELS

def assign_sentiment(rating):

    if rating <= 2:
        return 'Negative'

    else:
        return 'Non-negative'


# Apply the sentiment function
df_clean['sentiment'] = df_clean['rating'].apply(
    assign_sentiment
)

# Display the sentiment distribution
print("Sentiment distribution:")

print(df_clean['sentiment'].value_counts())

# Display percentages
print("\nSentiment percentages:")

display(
    df_clean['sentiment']
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)
```

Train the Logistic Regression model
Before training, we need to consider the severe class imbalance.

With only five non-negative reviews, a conventional train/test split can leave too few minority examples in the training or testing data. We will first inspect the available class counts.

```python

# STEP 11.1: CHECK CLASS COUNTS BEFORE SPLITTING

class_counts = df_clean['sentiment'].value_counts()

print(class_counts)

print("\nSmallest class size:", class_counts.min())
```

Create training and testing datasets
The following code uses a stratified split only if there are sufficient examples to place at least two examples from each class in the training and testing sets.

```python
from sklearn.model_selection import train_test_split

print("train_test_split imported successfully!")
```

```python
# Divide the data into training and testing sets

X = df_clean['cleaned_text']
y = df_clean['sentiment']

# Check the number of examples in each sentiment category
print("Sentiment distribution:")
print(y.value_counts())

# Split the dataset
if y.value_counts().min() >= 2:

    X_train, X_test, y_train, y_test = train_test_split(
        X,
        y,
        test_size=0.20,
        random_state=42,
        stratify=y
    )

    print("\nData successfully divided!")
    print("Training reviews:", len(X_train))
    print("Testing reviews:", len(X_test))

else:
    print("\nNot enough examples in one or more categories.")
    print("We need to adjust the classification approach.")
```

Even if the split runs successfully, the minority class is too small for reliable performance estimates. The test set may contain only one non-negative review.

Build a TF-IDF and Logistic Regression pipeline
A pipeline ensures that TF-IDF is fitted on training data only, reducing data leakage.

```python
# STEP 11.2: IMPORT MACHINE LEARNING LIBRARIES

from sklearn.pipeline import Pipeline

from sklearn.feature_extraction.text import TfidfVectorizer

from sklearn.feature_selection import SelectKBest, chi2

from sklearn.linear_model import LogisticRegression

print("All machine learning libraries imported successfully!")
```

```python

# STEP 11.3: CREATE THE MACHINE LEARNING PIPELINE

if y.value_counts().min() >= 2:

    model_pipeline = Pipeline([

        # Convert text into numerical TF-IDF features
        ('tfidf', TfidfVectorizer(
            max_features=1000,
            ngram_range=(1, 2),
            min_df=1,
            max_df=0.95
        )),

        # Select useful numerical features
        ('feature_selection', SelectKBest(
            score_func=chi2,
            k=100
        )),

        # Train Logistic Regression
        ('classifier', LogisticRegression(
            C=1.0,
            class_weight='balanced',
            max_iter=1000,
            random_state=42
        ))

    ])

    print("Machine learning pipeline created!")

else:

    print("Insufficient class examples for this experiment.")
```

```python

# STEP 11.4: TRAIN LOGISTIC REGRESSION

if y.value_counts().min() >= 2:

    model_pipeline.fit(X_train, y_train)

    print("Logistic Regression model trained successfully!")
```

Evaluate the model
We now test the trained model using the testing data.

```python

# STEP 12.1: PREDICT SENTIMENT

if y.value_counts().min() >= 2:

    y_pred = model_pipeline.predict(X_test)

    print("Actual labels:")
    print(list(y_test))

    print("\nPredicted labels:")
    print(list(y_pred))
```

```python
from sklearn.metrics import accuracy_score

print("Accuracy evaluation library imported successfully!")
```

```python

# STEP 12.2: CALCULATE MODEL ACCURACY

if y.value_counts().min() >= 2:

    accuracy = accuracy_score(y_test, y_pred)

    print("Model Accuracy:", round(accuracy, 4))

    print("Model Accuracy Percentage:",
          round(accuracy * 100, 2), "%")
```

Generate the classification report

```python
from sklearn.metrics import classification_report
print("classification report imported successfully!")
```

```python

# STEP 12.3: CLASSIFICATION REPORT

if y.value_counts().min() >= 2:

    print(
        classification_report(
            y_test,
            y_pred,
            labels=['Negative', 'Non-negative'],
            zero_division=0
        )
    )
```
